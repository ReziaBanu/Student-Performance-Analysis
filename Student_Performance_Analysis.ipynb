```python
data = {"students" : ['Arun', 'Banu', 'Charan', 'Divya', 'Ezhil'],
        " Maths_Marks" : ['85','75','90','80','95'],
        "Science_Marks" : ['78','88','92','75','85'],
        "Social_Marks" : ['90','82','85','88','92'] }
df = pd.DataFrame(data)
df
```

```python
## Matplot ##
plt.bar(x = df['students'], height = df[' Maths_Marks'].astype(int))
plt.title("comparision of Maths_Marks")
plt.xlabel("Names")
plt.ylabel("Marks in Mathematics")
plt.show()
```

```python
## Line_Plot ##

plt.plot(df['students'], df['Science_Marks'].astype(int), marker='*', linewidth=1, label = "Overall Science Marks" )
plt.title("comparision of Science_Marks")
plt.xlabel("Names")
plt.ylabel("Marks in Science")
plt.legend()
plt.grid()
plt.show()
```

```python
## Heat_Map ##
df = pd.DataFrame(data)
df[[' Maths_Marks', 'Science_Marks', 'Social_Marks']] = df[[' Maths_Marks', 'Science_Marks', 'Social_Marks']].astype(int)

df.set_index('students', inplace = True)

sns.heatmap(df, annot=True, cmap='Blues', linewidths=0.5)

plt.title("Students' Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Students")

plt.show()


```
